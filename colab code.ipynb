```python
# ============================================================
# SALES PERFORMANCE & REVENUE ANALYTICS - FULL COLAB NOTEBOOK
# Dataset: sales_performance_dataset_final.csv
# Outputs: KPIs, charts, CSV summaries, SQLite DB + SQL query results
# ============================================================

# -------------------------
# 0) Imports
# -------------------------
import os
import re
import sqlite3
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", 200)
pd.set_option("display.width", 200)

# -------------------------
# 1) Load Dataset
# -------------------------
# Option A: If you uploaded in Colab "Files"
DATA_PATH = "sales_performance_dataset_final.csv"

# Option B (Google Drive):
# from google.colab import drive
# drive.mount('/content/drive')
# DATA_PATH = "/content/drive/MyDrive/sales_performance_dataset_final.csv"

assert os.path.exists(DATA_PATH), f"File not found: {DATA_PATH}. Upload it to Colab or mount Drive."

df = pd.read_csv(DATA_PATH)

print("✅ Loaded:", DATA_PATH)
print("Shape:", df.shape)
display(df.head())

# -------------------------
# 2) Standardize Column Names (safe)
# -------------------------
def clean_col(col: str) -> str:
    col = col.strip().lower()
    col = re.sub(r"[^\w]+", "_", col)      # spaces/symbols -> _
    col = re.sub(r"_+", "_", col).strip("_")
    return col

df.columns = [clean_col(c) for c in df.columns]
print("Columns:", list(df.columns))

# -------------------------
# 3) Detect / Map Columns (robust for different schemas)
# -------------------------
def find_first_existing(candidates):
    for c in candidates:
        if c in df.columns:
            return c
    return None

col_order_id  = find_first_existing(["order_id", "orderid", "invoice_id", "transaction_id", "txn_id"])
col_date      = find_first_existing(["order_date", "date", "transaction_date", "invoice_date", "order_dt"])
col_region    = find_first_existing(["region", "state", "market", "sales_region"])
col_category  = find_first_existing(["product_category", "category", "sub_category", "product_cat"])
col_product   = find_first_existing(["product_name", "product", "item_name", "sku", "product_id"])
col_qty       = find_first_existing(["quantity", "qty", "units"])
col_price     = find_first_existing(["unit_price", "price", "unitprice"])
col_discount  = find_first_existing(["discount", "discount_pct", "disc", "discount_rate"])
col_sales     = find_first_existing(["sales_amount", "sales", "revenue", "amount", "total"])

print("Detected columns:")
print({
    "order_id": col_order_id, "order_date": col_date, "region": col_region,
    "category": col_category, "product": col_product, "qty": col_qty,
    "price": col_price, "discount": col_discount, "sales/revenue": col_sales
})

# Basic requirements: a date + revenue (or can compute revenue)
if col_date is None:
    raise ValueError("❌ Could not find a date column (expected something like order_date/date).")

# Parse dates
df[col_date] = pd.to_datetime(df[col_date], errors="coerce")
bad_dates = df[col_date].isna().sum()
print("Bad/NaT dates:", bad_dates)

# Ensure sales/revenue exists, else compute it from qty*price*(1-discount) if possible
if col_sales is None:
    if col_qty and col_price:
        # discount optional; if missing assume 0
        if col_discount is None:
            df["discount"] = 0.0
            col_discount = "discount"

        # If discount looks like 10 not 0.10, normalize
        # Heuristic: if median discount > 1, treat as percentage
        if pd.to_numeric(df[col_discount], errors="coerce").median() > 1:
            df[col_discount] = pd.to_numeric(df[col_discount], errors="coerce") / 100.0

        df["sales_amount"] = (
            pd.to_numeric(df[col_qty], errors="coerce").fillna(0) *
            pd.to_numeric(df[col_price], errors="coerce").fillna(0) *
            (1 - pd.to_numeric(df[col_discount], errors="coerce").fillna(0))
        )
        col_sales = "sales_amount"
        print("✅ Computed sales_amount from qty * unit_price * (1 - discount).")
    else:
        raise ValueError("❌ Could not find sales/revenue column and cannot compute it (missing qty/price).")

# Coerce revenue numeric
df[col_sales] = pd.to_numeric(df[col_sales], errors="coerce")
df = df.dropna(subset=[col_date, col_sales])

# -------------------------
# 4) Quick Data Quality Checks
# -------------------------
print("\n--- Missing values (top 15) ---")
display(df.isna().sum().sort_values(ascending=False).head(15))

print("\n--- Revenue sanity ---")
print("Min revenue:", df[col_sales].min(), "Max revenue:", df[col_sales].max())
print("Negative revenue rows:", (df[col_sales] < 0).sum())

# Remove negative revenue if present (common cleanup)
df = df[df[col_sales] >= 0].copy()

# -------------------------
# 5) Feature Engineering (Month, Year)
# -------------------------
df["year"] = df[col_date].dt.year
df["month"] = df[col_date].dt.to_period("M").astype(str)

display(df.head())

# -------------------------
# 6) KPI Calculations
# -------------------------
total_revenue = df[col_sales].sum()

if col_order_id:
    total_orders = df[col_order_id].nunique()
else:
    # fallback: treat each row as an order line -> approximate orders by row count
    total_orders = len(df)

aov = total_revenue / total_orders if total_orders else 0

kpis = pd.DataFrame({
    "metric": ["Total Revenue", "Total Orders", "Average Order Value (AOV)"],
    "value": [total_revenue, total_orders, aov]
})

print("\n✅ KPI Summary")
display(kpis)

# -------------------------
# 7) Aggregations for Analysis
# -------------------------
monthly_revenue = (
    df.groupby("month", as_index=False)[col_sales]
      .sum()
      .rename(columns={col_sales: "monthly_revenue"})
      .sort_values("month")
)

monthly_revenue["prev_month_revenue"] = monthly_revenue["monthly_revenue"].shift(1)
monthly_revenue["mom_growth_pct"] = np.where(
    monthly_revenue["prev_month_revenue"].isna() | (monthly_revenue["prev_month_revenue"] == 0),
    np.nan,
    (monthly_revenue["monthly_revenue"] - monthly_revenue["prev_month_revenue"]) / monthly_revenue["prev_month_revenue"] * 100
)

# Region revenue (if available)
region_revenue = None
if col_region:
    region_revenue = (
        df.groupby(col_region, as_index=False)[col_sales]
          .sum()
          .rename(columns={col_sales: "region_revenue"})
          .sort_values("region_revenue", ascending=False)
    )

# Category performance (if available)
category_revenue = None
if col_category:
    category_revenue = (
        df.groupby(col_category, as_index=False)[col_sales]
          .sum()
          .rename(columns={col_sales: "category_revenue"})
          .sort_values("category_revenue", ascending=False)
    )

# Top products (if available)
top_products = None
if col_product:
    top_products = (
        df.groupby(col_product, as_index=False)[col_sales]
          .sum()
          .rename(columns={col_sales: "product_revenue"})
          .sort_values("product_revenue", ascending=False)
          .head(10)
    )

print("\n✅ Monthly Revenue (with MoM Growth)")
display(monthly_revenue.head(12))

if region_revenue is not None:
    print("\n✅ Revenue by Region")
    display(region_revenue)

if category_revenue is not None:
    print("\n✅ Revenue by Category")
    display(category_revenue)

if top_products is not None:
    print("\n✅ Top 10 Products by Revenue")
    display(top_products)

# -------------------------
# 8) Charts (Matplotlib)
# -------------------------
# Monthly revenue trend
plt.figure(figsize=(12, 5))
plt.plot(monthly_revenue["month"], monthly_revenue["monthly_revenue"], marker="o")
plt.xticks(rotation=45, ha="right")
plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.tight_layout()
plt.show()

# MoM growth
plt.figure(figsize=(12, 5))
plt.plot(monthly_revenue["month"], monthly_revenue["mom_growth_pct"], marker="o")
plt.xticks(rotation=45, ha="right")
plt.title("Month-over-Month (MoM) Growth %")
plt.xlabel("Month")
plt.ylabel("MoM Growth (%)")
plt.axhline(0, linewidth=1)
plt.tight_layout()
plt.show()

# Region bar chart
if region_revenue is not None:
    plt.figure(figsize=(10, 5))
    plt.bar(region_revenue[col_region].astype(str), region_revenue["region_revenue"])
    plt.xticks(rotation=45, ha="right")
    plt.title("Revenue by Region")
    plt.xlabel("Region")
    plt.ylabel("Revenue")
    plt.tight_layout()
    plt.show()

# Category bar chart
if category_revenue is not None:
    plt.figure(figsize=(10, 5))
    plt.bar(category_revenue[col_category].astype(str), category_revenue["category_revenue"])
    plt.xticks(rotation=45, ha="right")
    plt.title("Revenue by Category")
    plt.xlabel("Category")
    plt.ylabel("Revenue")
    plt.tight_layout()
    plt.show()

# Top products bar chart
if top_products is not None:
    plt.figure(figsize=(12, 5))
    plt.bar(top_products[col_product].astype(str), top_products["product_revenue"])
    plt.xticks(rotation=75, ha="right")
    plt.title("Top 10 Products by Revenue")
    plt.xlabel("Product")
    plt.ylabel("Revenue")
    plt.tight_layout()
    plt.show()

# -------------------------
# 9) Export Summaries (for Dashboard + GitHub)
# -------------------------
os.makedirs("outputs", exist_ok=True)

kpis.to_csv("outputs/kpi_summary.csv", index=False)
monthly_revenue.to_csv("outputs/monthly_revenue.csv", index=False)

if region_revenue is not None:
    region_revenue.to_csv("outputs/region_revenue.csv", index=False)

if category_revenue is not None:
    category_revenue.to_csv("outputs/category_revenue.csv", index=False)

if top_products is not None:
    top_products.to_csv("outputs/top_products.csv", index=False)

print("✅ Exported outputs to /content/outputs/")
print(os.listdir("outputs"))

# -------------------------
# 10) Create SQLite DB + Run Resume-Grade SQL Queries
# -------------------------
db_path = "outputs/sales_analytics.db"
conn = sqlite3.connect(db_path)

# Store full fact table
df_sql = df.copy()
# SQLite likes simple column names; already standardized above
df_sql.to_sql("sales", conn, if_exists="replace", index=False)

# Helper to run SQL and show results
def run_sql(q, name=None):
    out = pd.read_sql_query(q, conn)
    if name:
        out.to_csv(f"outputs/{name}.csv", index=False)
    return out

# Total revenue
q_total_revenue = f"""
SELECT SUM({col_sales}) AS total_revenue
FROM sales;
"""
display(run_sql(q_total_revenue, "sql_total_revenue"))

# Average order value (AOV) - if order_id exists, else approximate by rows
if col_order_id:
    q_aov = f"""
    SELECT
        SUM({col_sales}) / COUNT(DISTINCT {col_order_id}) AS aov
    FROM sales;
    """
else:
    q_aov = f"""
    SELECT
        AVG({col_sales}) AS aov_approx_per_row
    FROM sales;
    """
display(run_sql(q_aov, "sql_aov"))

# Monthly revenue
q_monthly = f"""
SELECT
    SUBSTR({col_date}, 1, 7) AS month,
    SUM({col_sales}) AS monthly_revenue
FROM sales
GROUP BY 1
ORDER BY 1;
"""
monthly_sql = run_sql(q_monthly, "sql_monthly_revenue")
display(monthly_sql.head(12))

# MoM growth (SQLite window function supported in modern SQLite)
q_mom = f"""
WITH m AS (
  SELECT
      SUBSTR({col_date}, 1, 7) AS month,
      SUM({col_sales}) AS revenue
  FROM sales
  GROUP BY 1
)
SELECT
    month,
    revenue,
    LAG(revenue) OVER (ORDER BY month) AS prev_month,
    CASE
      WHEN LAG(revenue) OVER (ORDER BY month) IS NULL OR LAG(revenue) OVER (ORDER BY month) = 0
      THEN NULL
      ELSE ROUND((revenue - LAG(revenue) OVER (ORDER BY month)) * 100.0 / LAG(revenue) OVER (ORDER BY month), 2)
    END AS mom_growth_pct
FROM m
ORDER BY month;
"""
display(run_sql(q_mom, "sql_mom_growth").head(12))

# Top products
if col_product:
    q_top_products = f"""
    SELECT
        {col_product} AS product,
        SUM({col_sales}) AS revenue
    FROM sales
    GROUP BY 1
    ORDER BY revenue DESC
    LIMIT 10;
    """
    display(run_sql(q_top_products, "sql_top_products"))

# Region revenue
if col_region:
    q_region = f"""
    SELECT
        {col_region} AS region,
        SUM({col_sales}) AS revenue
    FROM sales
    GROUP BY 1
    ORDER BY revenue DESC;
    """
    display(run_sql(q_region, "sql_region_revenue"))

# Underperforming categories (avg sales)
if col_category:
    q_underperform = f"""
    SELECT
        {col_category} AS category,
        AVG({col_sales}) AS avg_sale
    FROM sales
    GROUP BY 1
    ORDER BY avg_sale ASC;
    """
    display(run_sql(q_underperform, "sql_underperforming_categories"))

conn.close()
print(f"✅ SQLite DB saved: {db_path}")

# -------------------------
# 11) Auto-Generate "Insights" (simple, resume-friendly)
# -------------------------
insights = []

# Seasonal peak month
peak_row = monthly_revenue.loc[monthly_revenue["monthly_revenue"].idxmax()]
insights.append(f"Peak revenue month was {peak_row['month']} with revenue {peak_row['monthly_revenue']:.2f}.")

# Highest region
if region_revenue is not None and len(region_revenue) > 0:
    top_r = region_revenue.iloc[0]
    insights.append(f"Top region by revenue: {top_r[col_region]} ({top_r['region_revenue']:.2f}).")

# Highest category
if category_revenue is not None and len(category_revenue) > 0:
    top_c = category_revenue.iloc[0]
    insights.append(f"Top category by revenue: {top_c[col_category]} ({top_c['category_revenue']:.2f}).")

# Concentration (top 10 products share)
if top_products is not None and len(top_products) > 0:
    top10_share = top_products["product_revenue"].sum() / total_revenue * 100
    insights.append(f"Top 10 products contribute approximately {top10_share:.2f}% of total revenue.")

insights_df = pd.DataFrame({"insight": insights})
insights_df.to_csv("outputs/insights_summary.csv", index=False)

print("\n✅ Auto Insights")
display(insights_df)

# -------------------------
# 12) (Optional) Zip outputs for GitHub upload/download
# -------------------------
import shutil
shutil.make_archive("sales_analytics_outputs", "zip", "outputs")
print("✅ Created: sales_analytics_outputs.zip")
```
